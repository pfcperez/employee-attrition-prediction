### Feature engineering

```python
data.columns
```

```python
df_processed = data.copy()
```

Dropping columns where the value is the same for all of them
- Over18
- StandardHours

```python
df_processed.drop(['Over18','StandardHours','EmployeeNumber'],axis=1, inplace=True)
```

Mapping categorical variables

```python
df_processed.select_dtypes(exclude=['number'])
```

```python
df_processed['Attrition'] = df_processed['Attrition'].map({'No': 0, 'Yes': 1})
df_processed['OverTime'] = df_processed['OverTime'].map({'No': 0, 'Yes': 1})
df_processed['Gender'] = df_processed['Gender'].map({'Female': 0, 'Male': 1})

travel_map = {'Non-Travel': 0, 'Travel_Rarely': 1, 'Travel_Frequently': 2}
df_processed['BusinessTravel'] = df_processed['BusinessTravel'].map(travel_map)


columnas_nominales = ['Department', 'EducationField', 'JobRole', 'MaritalStatus']


df_processed = pd.get_dummies(df_processed, columns=columnas_nominales, drop_first=True, dtype=int)
```

```python
print(df_processed.info())
```

```python
df_processed.head()
```

### Random forest classifier

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split,GridSearchCV
from sklearn.metrics import classification_report, accuracy_score
```

```python
X = df_processed.drop(['Attrition'],axis=1)
y = df_processed['Attrition']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=.30, random_state=45)
```

```python
rf_cl = RandomForestClassifier(n_estimators=100, max_depth=8, random_state=42)
```

```python
rf_cl.fit(X_train, y_train)
```

```python
y_pred = rf_cl.predict(X_test)
```

```python
print(f"Accuracy: {accuracy_score(y_test, y_pred):.2f}")
print(classification_report(y_test, y_pred, target_names=['No', 'Yes']))
```

Grid search

```python
rf_grid = RandomForestClassifier(random_state=42)
```

```python
parameters = {
    'n_estimators':[100,200,300],
    'max_depth':[5,8,10],
    'min_samples_split':[3,6,9],
    'min_samples_leaf':[2,4,6],
    'max_features':['sqrt', 'log2']
}
```

```python
grid_search = GridSearchCV(
    estimator=rf_grid, 
    param_grid=parameters, 
    cv=5,           
    scoring='accuracy', 
    n_jobs=-1,      
    verbose=1      
)
```

```python
grid_search.fit(X_train, y_train)
```

```python
print("Best Parameters:", grid_search.best_params_)
print("Best CV Score:", grid_search.best_score_)
```

```python
best_model = grid_search.best_estimator_
test_accuracy = best_model.score(X_test, y_test)
print("Test Accuracy:", test_accuracy)
```

### XGBoost 

```python
from xgboost import XGBClassifier
```

```python
xgboost_model = XGBClassifier(
    n_estimators = 100,
    learning_rate=0.1,
    max_depth = 10
)
```

```python
xgboost_model.fit(X_train, y_train)
```

```python
y_pred = xgboost_model.predict(X_test)
print("Accuracy:", accuracy_score(y_test, y_pred))
```
